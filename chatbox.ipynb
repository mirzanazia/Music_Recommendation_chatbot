```python
df = pd.read_csv("songs_datasets.csv")

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
df = df.dropna(subset=["track_name", "artist(s)_name"])

df["track_name"] = df["track_name"].astype(str).str.strip()
df["artist(s)_name"] = df["artist(s)_name"].astype(str).str.strip()

print("Data cleaning completed successfully!")
print("Number of songs after cleaning:", len(df))
```

```python
df = df.drop_duplicates(subset=["track_name", "artist(s)_name"])

print("Duplicate songs removed successfully!")
print("Number of songs after removing duplicates:", len(df))
```

```python
def recommend_songs(song_name, number_of_recommendations=5):
    
    # Find the selected song
    result = df[
        df["track_name"].str.lower().str.strip()
        == song_name.lower().strip()
    ]
    
    # Check if the song exists
    if result.empty:
        print("Song not found in the dataset.")
        return
    
    # Get the selected song
    selected_song = result.iloc[0]
    
    # Get the artist name
    artist = selected_song["artist(s)_name"]
    
    print("Selected Song:", selected_song["track_name"])
    print("Artist:", artist)
    
    # Find other songs by the same artist
    recommendations = df[
        (df["artist(s)_name"].str.lower() == str(artist).lower())
        & (df["track_name"].str.lower() != song_name.lower().strip())
    ]
    
    # Display recommendations
    if recommendations.empty:
        print("No other songs by this artist were found.")
    
    else:
        print("\nRecommended Songs:")
        
        for _, row in recommendations.head(number_of_recommendations).iterrows():
            print(
                "-",
                row["track_name"],
                "by",
                row["artist(s)_name"]
            )
```

```python
def song_details(song_name):
    
    result = df[
        df["track_name"].str.lower().str.strip()
        == song_name.lower().strip()
    ]
    
    if result.empty:
        print("Song not found in the dataset.")
        return
    
    song = result.iloc[0]
    
    print("Song Details")
    print("-------------------------")
    print("Song:", song["track_name"])
    print("Artist:", song["artist(s)_name"])
    print("Released Year:", song["released_year"])
    print("Streams:", song["streams"])
```

```python
song_details("Cruel Summer")
```

```python
song_name = "Cruel Summer"

print("MUSIC RECOMMENDATION CHATBOT")
print("============================")
print()

song_details(song_name)

print()
print("============================")
print()

recommend_songs(song_name)
```
